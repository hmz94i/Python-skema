## Exercise 3.1: Odd or even (homework)

```text
WHAT THE PROGRAM MUST DO
    Ask the user for a number N, then say for every number from 1 to N whether it is
    odd or even.

ANSWER THESE FIRST, in comments at the top of your file
    1. What goes in?
    2. What happens to it?
    3. What comes out?
    4. What should happen if the user types 0, a negative number, or 5000?
       Decide the three behaviours before writing anything.

WHAT THE AI CANNOT KNOW
    Your three decisions. An assistant asked for "odd or even from 1 to N" will produce
    a program that behaves absurdly on 0 and on -4, and will happily print five thousand
    lines. Those are your calls, not its.

CHECK IT YOURSELF
    Run it with 6. You should see three odd and three even. Count them.
    Then run it with your three edge cases and confirm each does what you decided.

DELIVERABLE
    This file, with your comments and your code.
```

````python
# 1. In: one whole number N typed by the user.
# 2. Process: the program checks N first. If N is fine, it goes through every number from 1 to N and uses the remainder of the division by 2 to know if it is odd or even.
# 3. Out: one line per number, saying "odd" or "even". Or one message if N is 0, negative or too large.
# 4. What happens on 0, on a negative number, on a very large number:
#    - 0: a message says there is nothing to check, because there is no number between 1 and 0.
#    - Negative: a message asks for a positive number, for the same reason.
#    - Above 100: a message says the maximum is 100, because more lines would be unreadable.

# Your code below

# Ask for N and convert it to a whole number. The maximum is in a variable, easy to change
n = int(input("Up to which number? "))
maximum = 100

# Check the three special cases first
if n == 0:
    print("Nothing to check, please type a number from 1.")
elif n < 0:
    print("Please type a positive number.")
elif n > maximum:
    print("Too many numbers, the maximum is", maximum)
else:
    # Normal case: for every number from 1 to n, look at the remainder of the division by 2
    for number in range(1, n + 1):
        if number % 2 == 0:
            print(number, "is even")
        else:
            print(number, "is odd")

# CHECK IT YOURSELF
# 6: 1, 3, 5 are odd and 2, 4, 6 are even. Three odd and three even. Correct.
# 0: "Nothing to check, please type a number from 1." As decided.
# -4: "Please type a positive number." As decided.
# 5000: "Too many numbers, the maximum is 100". As decided, no 5000 lines.
````

## Exercise 5.2: Repeating until something changes

```text
WHAT THE PROGRAM MUST DO
    Keep asking the user something until a condition you define is met, then display a
    summary of what happened during the loop.

ANSWER THESE FIRST, in comments at the top of your file
    1. What goes in?
    2. What happens to it?
    3. What comes out?
    4. What is your stop condition, what is your maximum number of attempts, and what
       does your summary contain?

WHAT THE AI CANNOT KNOW
    Your stop condition and your safety limit. An assistant asked for a while loop will
    write one that can run for ever if the user never gives the expected answer. Decide
    how many attempts you allow, and what your program does when that limit is reached.

    Accepting "Yes", "yes" and " yes " as the same answer is your decision too. Make it
    and write it down.

CHECK IT YOURSELF
    Run it and never give the expected answer. If your program is still running after
    your stated maximum, it is wrong. Then run it and answer with capitals and extra
    spaces.

DELIVERABLE
    This file, with your comments and your code.
```

````python
# 1. In: the answers typed by the guest, one per attempt.
# 2. Process: the program asks the same question again while the answer is not "yes" and attempts are left. Each answer is cleaned with strip() and lower().
# 3. Out: a summary with the number of attempts used, and whether the order is confirmed or cancelled.
# 4. My stop condition, my attempt limit, my summary:
#    - Stop condition: the guest answers "yes". "Yes", "YES" and " yes " count as the same answer.
#    - Limit: 3 attempts. After 3 answers without "yes", the order is cancelled.
#    - Summary: the number of attempts used, then "confirmed" or "cancelled".

# Your code below

# The limit is in a variable, easy to change. The counter starts at 0, the answer starts empty
max_attempts = 3
attempts = 0
answer = ""

# Ask again while the answer is not "yes" AND attempts are left
while answer != "yes" and attempts < max_attempts:
    answer = input("Do you confirm your room service order? (yes/no) ")
    # Clean the answer: remove the spaces at both ends, put everything in lower case
    answer = answer.strip().lower()
    # One more attempt used. Without this line, the loop could run for ever
    attempts = attempts + 1

# Summary of what happened in the loop
print("Attempts used:", attempts, "out of", max_attempts)
if answer == "yes":
    print("Order confirmed.")
else:
    print("No confirmation after", max_attempts, "attempts: order cancelled.")

# CHECK IT YOURSELF
# Never the expected answer (no, no, no): the loop stops after 3 attempts. "Attempts used: 3 out of 3", order cancelled. Correct.
# no, then "   YES   " with capitals and spaces: accepted at the 2nd attempt. "Attempts used: 2 out of 3", order confirmed. Correct.
# An empty answer (just Enter) counts as one attempt, so the loop still stops at 3.
````
