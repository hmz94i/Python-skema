````python
# 1. In: two answers typed by the guest: the number of meals per day (a number) and their allergies (text).
# 2. Process: the number of meals is typed as text, so it is converted into a whole number with int(). Then both answers are put into one sentence.
# 3. Out: one sentence for the kitchen with the number of meals and the allergies.
# 4. My two fields, and what I would do with them: meals per day and allergies, for a hotel room service. The kitchen knows how many meals to prepare and what to avoid.

# Your code below

# Ask the two questions. input() always gives back text
meals_text = input("How many meals per day do you want? ")
allergies = input("Any allergies? ")

# Convert the number of meals from text to a whole number
meals = int(meals_text)

# Build one sentence with both answers
print("Room service: " + str(meals) + " meals per day. Allergies: " + allergies + ".")

# Add 1 to the number to show it is now a real number
print("With one extra snack, that makes " + str(meals + 1) + " deliveries per day.")

# CHECK IT YOURSELF
# Normal case (3, peanuts): "Room service: 3 meals per day. Allergies: peanuts." then 4 deliveries. Correct.
# Empty line for the meals: crash. ValueError: invalid literal for int() with base 10: ''
# A single space for the meals: crash. ValueError: invalid literal for int() with base 10: ' '
# Text for the meals ("five"): crash. ValueError: invalid literal for int() with base 10: 'five'
# In all three cases the error is on line 13, meals = int(meals_text): int() only accepts digits. Not fixed yet, as asked.
````

````python
# 1. In: two numbers typed by the user.
# 2. Process: both answers are converted from text to decimal numbers with float(). Then the program computes the sum, the difference, the product and the division.
# 3. Out: four lines, one per operation.
# 4. What happens when the second number is zero, and why: the division line shows a message instead of a result. Only the division is impossible with 0. The three other results are correct, so the program still shows them.

# Your code below

# Ask for the two numbers and convert them from text to decimal numbers
first = float(input("First number: "))
second = float(input("Second number: "))

# The three operations that always work
print("Sum:", first + second)
print("Difference:", first - second)
print("Product:", first * second)

# The division: check first if the second number is 0
if second == 0:
    print("Division: impossible, division by zero")
else:
    print("Division:", first / second)

# CHECK IT YOURSELF
# 7 and 2: in my head 7 / 2 = 3.5. The program shows 3.5. Correct.
# It would show 3 with // instead of /: // is the whole division, it drops the decimals.
# 7 and 0: sum 7.0, difference 7.0, product 0.0, and the division line shows the message. No crash.
````

## Exercise 2.1: Transforming text (homework)

```text
WHAT THE PROGRAM MUST DO
    Ask the user for a sentence, then display four different transformations of it.

ANSWER THESE FIRST, in comments at the top of your file
    1. What goes in?
    2. What happens to it?
    3. What comes out?
    4. Which four transformations did you choose, and in what situation would each of
       them be useful? One line each.

    to see what is available, then choose, then justify.

    A transformation that produces the same thing as another one does not count as two.

    For each of your four results, say in a comment whether it is what you expected.
```

````python
# 1. In: one sentence typed by the user.
# 2. Process: the program first removes the spaces at both ends, then applies three other transformations to the cleaned sentence.
# 3. Out: four lines, one per transformation, between brackets so the spaces are visible.
# 4. My four transformations, and when each is useful:
#    - strip(): remove the spaces at both ends, to clean cells copied from an Excel export.
#    - lower(): everything in lower case, to compare two email addresses written differently.
#    - title(): a capital at the start of each word, to clean up client names in a file.
#    - replace(" ", "_"): spaces become underscores, to build a file name.

# Your code below

# Ask for a sentence
sentence = input("Type a sentence: ")

# Remove the spaces at both ends. The other transformations start from this clean version
cleaned = sentence.strip()

# Show the four transformations. The brackets show if spaces are left at the ends
print("[" + cleaned + "]")
print("[" + cleaned.lower() + "]")
print("[" + cleaned.title() + "]")
print("[" + cleaned.replace(" ", "_") + "]")

# CHECK IT YOURSELF, with "   i am the Best of all time   " (spaces at both ends, a capital in the middle)
# - strip: [i am the Best of all time]. Expected: the spaces at both ends are gone, the inside is unchanged.
# - lower: [i am the best of all time]. Expected: the capital B in the middle becomes b.
# - title: [I Am The Best Of All Time]. Expected, but "Of" also gets a capital: title() does not know grammar.
# - replace: [i_am_the_Best_of_all_time]. Expected. The end spaces did not become "_" because strip ran first.
````
